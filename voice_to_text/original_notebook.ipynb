```python
import pyaudio
import wave

FORMAT = pyaudio.paInt16
CHANNELS = 1
RATE = 16000
CHUNK = 1024

RECORD_SECONDS = 5
OUTPUT_FILE = "audio.wav"

audio = pyaudio.PyAudio()

print("Opening microphone...")

stream = audio.open(
    format=FORMAT,
    channels=CHANNELS,
    rate=RATE,
    input=True,
    frames_per_buffer=CHUNK
)

print("\n🎤 Recording started!")
print("Speak now...")

frames = []

for i in range(0, int(RATE / CHUNK * RECORD_SECONDS)):
    data = stream.read(CHUNK)
    frames.append(data)

print("Recording finished!")

stream.stop_stream()
stream.close()

sample_width = audio.get_sample_size(FORMAT)
audio.terminate()

with wave.open(OUTPUT_FILE, "wb") as wf:
    wf.setnchannels(CHANNELS)
    wf.setsampwidth(sample_width)
    wf.setframerate(RATE)
    wf.writeframes(b"".join(frames))

print("Audio saved as:", OUTPUT_FILE)
```

```python
import pyaudio
import wave
import whisper
import threading

# ====================================
# STEP 1: RECORD AUDIO
# ====================================

FORMAT = pyaudio.paInt16
CHANNELS = 1
RATE = 16000
CHUNK = 1024

OUTPUT_FILE = "audio.wav"

audio = pyaudio.PyAudio()

stream = audio.open(
    format=FORMAT,
    channels=CHANNELS,
    rate=RATE,
    input=True,
    frames_per_buffer=CHUNK
)

print("\n🎤 Recording started!")
print("Speak now...")
print("⏹️ Press ENTER to stop recording.")

frames = []

# Stop control
stop_recording = False


def stop():
    global stop_recording
    input()
    stop_recording = True


# Separate thread for ENTER key
thread = threading.Thread(target=stop)
thread.daemon = True
thread.start()


# ====================================
# RECORD UNTIL ENTER IS PRESSED
# ====================================

while not stop_recording:
    data = stream.read(CHUNK, exception_on_overflow=False)
    frames.append(data)

print("\n⏹️ Recording stopped!")

stream.stop_stream()
stream.close()

sample_width = audio.get_sample_size(FORMAT)
audio.terminate()


# ====================================
# SAVE AUDIO
# ====================================

with wave.open(OUTPUT_FILE, "wb") as wf:
    wf.setnchannels(CHANNELS)
    wf.setsampwidth(sample_width)
    wf.setframerate(RATE)
    wf.writeframes(b"".join(frames))

print("💾 Audio saved as:", OUTPUT_FILE)


# ====================================
# STEP 2: LOAD WHISPER
# ====================================

print("\n⏳ Loading Whisper model...")

model = whisper.load_model("base")


# ====================================
# STEP 3: TRANSCRIBE
# ====================================

print("📝 Converting speech to text...")

result = model.transcribe(
    OUTPUT_FILE,
    fp16=False
)


# ====================================
# STEP 4: DISPLAY RESULT
# ====================================

print("\n================================")
print("       TRANSCRIBED TEXT")
print("================================")

print(result["text"])

print("================================")
```

```python
import pyaudio
import wave
import whisper
import time

# ====================================
# SETTINGS
# ====================================

FORMAT = pyaudio.paInt16
CHANNELS = 1
RATE = 16000
CHUNK = 1024

OUTPUT_FILE = "audio.wav"

# Silence settings
SILENCE_LIMIT = 5        # 5 seconds silence = STOP
START_TIMEOUT = 10       # voice start karne ke liye 10 sec

# ====================================
# RECORDING SETUP
# ====================================

audio = pyaudio.PyAudio()

stream = audio.open(
    format=FORMAT,
    channels=CHANNELS,
    rate=RATE,
    input=True,
    frames_per_buffer=CHUNK
)

print("\n🎤 Recording started!")
print("Speak now...")
print("⏸️ Short pauses are allowed.")
print("🛑 Recording will stop after 5 seconds of silence.")

frames = []

silence_start = None
voice_started = False
start_time = time.time()

# ====================================
# RECORDING LOOP
# ====================================

while True:

    data = stream.read(
        CHUNK,
        exception_on_overflow=False
    )

    frames.append(data)

    # Convert audio data to numbers
    import audioop

    volume = audioop.rms(data, 2)

    # -----------------------------
    # VOICE DETECTED
    # -----------------------------

    if volume > 500:

        voice_started = True
        silence_start = None

        print("🎤", end="", flush=True)

    # -----------------------------
    # SILENCE DETECTED
    # -----------------------------

    else:

        if voice_started:

            if silence_start is None:
                silence_start = time.time()

            silence_duration = time.time() - silence_start

            print(
                f"\r⏸️ Silence: {silence_duration:.1f} sec",
                end="",
                flush=True
            )

            # Stop after 5 sec silence
            if silence_duration >= SILENCE_LIMIT:

                print("\n\n🛑 Recording stopped automatically!")
                break

        else:

            # No voice at beginning
            if time.time() - start_time > START_TIMEOUT:

                print("\n❌ No voice detected.")
                break


# ====================================
# CLOSE RECORDING
# ====================================

stream.stop_stream()
stream.close()

sample_width = audio.get_sample_size(FORMAT)

audio.terminate()


# ====================================
# SAVE AUDIO
# ====================================

with wave.open(OUTPUT_FILE, "wb") as wf:

    wf.setnchannels(CHANNELS)
    wf.setsampwidth(sample_width)
    wf.setframerate(RATE)

    wf.writeframes(b"".join(frames))

print("💾 Audio saved as:", OUTPUT_FILE)


# ====================================
# LOAD WHISPER
# ====================================

print("\n⏳ Loading Whisper model...")

model = whisper.load_model("base")


# ====================================
# TRANSCRIBE
# ====================================

print("📝 Converting speech to text...")

result = model.transcribe(
    OUTPUT_FILE,
    fp16=False
)


# ====================================
# DISPLAY RESULT
# ====================================

print("\n================================")
print("       TRANSCRIBED TEXT")
print("================================")

print(result["text"])

print("================================")
```
